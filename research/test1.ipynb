```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()
```

```python
llm=ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.4
)
```

```python
response = llm.invoke(
    "Hello! Explain what Agentic AI is."
)
display(Markdown(response.content))
```
